# Sideview glancing benchmark

Shell for timing and accuracy sweeps over energy, glancing angle, and grid size.

```python
import time

import jax
import jax.numpy as jnp

from wide_angle_propagation.propagation_methods import simulate_bidirectional_wpm_1d

jax.config.update("jax_enable_x64", True)

def timed_call(fn, *args, **kwargs):
    t0 = time.perf_counter()
    result = fn(*args, **kwargs)
    jax.tree.map(lambda x: x.block_until_ready() if hasattr(x, "block_until_ready") else x, result)
    return result, time.perf_counter() - t0

energy = 100e3
n = 128
dx = 0.1
dz = 0.5
wave = jnp.ones(n, dtype=jnp.complex128)
potential = jnp.zeros((4, n))

_, elapsed = timed_call(simulate_bidirectional_wpm_1d, potential, wave, dx, dz, energy, n_bins=16)
elapsed
```
